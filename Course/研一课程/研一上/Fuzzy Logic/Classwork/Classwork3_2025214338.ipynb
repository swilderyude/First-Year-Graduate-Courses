```python
# Fuzzy Set Operations + Graphs
# A ∪ (B ∪ C) = (A ∪ B) ∪ C
# (A ∩ (B ∪ C) =(A ∩ B)∪(A ∩ C)
import numpy as np
import matplotlib.pyplot as plt

# ----- Data -----
U = np.array([1.0,1.5,2.0,2.5, 3.0])
A = np.array([1.0, 0.65, 0.4, 0.35, 0.0])
B = np.array([0.0, 0.25, 0.6, 0.25, 0.1])
C = np.array([0.5,0.25,0.0,0.25,0.5])

# ----- Operations -----
# A_union_B = np.maximum(A, B)  # A ∪ B
# A_inter_B = np.minimum(A, B)  # A ∩ B
# A_comp = 1 - A  # Ā
# B_comp = 1 - B  # B̄
A_union_B = np.maximum(A, B)  
B_union_C = np.maximum(B,C)
A_inter_B = np.minimum(A, B)  
A_inter_C = np.minimum(A, C)  

resultleft1 = np.maximum(A, B_union_C)
resultright1 = np.maximum(A_union_B,C)

resultleft2 = np.minimum(A, B_union_C)
resultright2 = np.maximum(A_inter_B,A_inter_C)

# ----- Numeric results -----
print("Universe U       :", U)
print("A                :", A)
print("B                :", B)
print("resultleft1 A ∪ (B ∪ C)     :", resultleft1)
print("resultright1 (A ∪ B) ∪ C    :", resultright1)

print("resultleft2 (A ∩ (B ∪ C)     :", resultleft2)
print("resultright2 (A ∩ B)∪(A ∩ C)     :", resultright2)

# ----- Plot 1: Membership functions A and B -----
plt.figure()
plt.plot(U, A, marker="o", label="A (membership)")
plt.plot(U, B, marker="s", label="B (membership)")
plt.title("Fuzzy Sets A and B: Membership Functions")
plt.xlabel("Universe U")
plt.ylabel("Membership μ(x)")
plt.ylim(-0.05, 1.05)
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()

# ----- Plot 2: Result1 (A ∪ (B ∪ C) = (A ∪ B) ∪ C) -----
plt.figure()
plt.plot(U, resultleft1, marker="^", label="A ∪ (B ∪ C)")
plt.plot(U, resultright1, marker="v", label="(A ∪ B) ∪ C")
plt.title("Result1 (A ∪ (B ∪ C) = (A ∪ B) ∪ C)")
plt.xlabel("Universe U")
plt.ylabel("Membership μ(x)")
plt.ylim(-0.05, 1.05)
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()

# ----- Plot 3: Result2 ((A ∩ (B ∪ C) =(A ∩ B)∪(A ∩ C)) -----
plt.figure()
plt.plot(U, resultleft2, marker="^", label="(A ∩ (B ∪ C)")
plt.plot(U, resultright2, marker="v", label="(A ∩ B)∪(A ∩ C)")
plt.title("Result2 ((A ∩ (B ∪ C) =(A ∩ B)∪(A ∩ C))")
plt.xlabel("Universe U")
plt.ylabel("Membership μ(x)")
plt.ylim(-0.05, 1.05)
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()
```

```python
# Fuzzy Set Operations + Graphs
# A ∩ B̄, B ∩ Ā
import numpy as np
import matplotlib.pyplot as plt

# ----- Data -----
U = np.array([2, 3, 4, 5, 6])
A = np.array([0.1, 0.6, 0.4, 0.3, 0.8])
B = np.array([0.5, 0.8, 0.4, 0.6, 0.4])

# ----- Operations -----
A_inter_B = np.minimum(A, B)  # A ∩ B
A_comp = 1 - A  # Ā
B_comp = 1 - B  # B̄

result1 = np.minimum(A, B_comp)
result2 = np.minimum(B, A_comp)

# ----- Numeric results -----
print("Universe U       :", U)
print("A                :", A)
print("B                :", B)
print("result1 A ∩ B̄     :", result1)
print("result2 B ∩ Ā     :", result2)

# ----- Plot 1: Membership functions A and B -----
plt.figure()
plt.plot(U, A, marker="o", label="A (membership)")
plt.plot(U, B, marker="s", label="B (membership)")
plt.title("Fuzzy Sets A and B: Membership Functions")
plt.xlabel("Universe U")
plt.ylabel("Membership μ(x)")
plt.ylim(-0.05, 1.05)
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()

# ----- Plot 2: Result (A ∩ B̄, B ∩ Ā) -----
plt.figure()
plt.plot(U, result1, marker="^", label="A ∩ B̄")
plt.plot(U, result2, marker="v", label="B ∩ Ā")
plt.title("Result (A ∩ B̄, B ∩ Ā)")
plt.xlabel("Universe U")
plt.ylabel("Membership μ(x)")
plt.ylim(-0.05, 1.05)
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()
```

```python
# Fuzzy Set Operations + Graphs
# (A ∪ B)_comp = Ā ∩ B̄
# (A ∩ B)_comp = Ā ∪ B̄
import numpy as np
import matplotlib.pyplot as plt

# ----- Data -----
U = np.array([10, 15, 20, 25])
A = np.array([0.8, 0.3, 0.6, 0.2])
B = np.array([0.4, 0.2, 0.9, 0.1])

# ----- Operations -----
A_comp = 1 - A  # Ā
B_comp = 1 - B  # B̄
A_union_B = np.maximum(A, B)
A_inter_B = np.minimum(A, B)


resultleft1 = 1 - A_union_B
resultright1 = np.minimum(A_comp, B_comp)

resultleft2 = 1 - A_inter_B
resultright2 = np.maximum(A_comp, B_comp)

# ----- Numeric results -----
print("Universe U       :", U)
print("A                :", A)
print("B                :", B)
print("resultleft1 (A ∪ B)_comp     :", resultleft1)
print("resultright1 Ā ∩ B̄    :", resultright1)

print("resultleft2 (A ∩ B)_comp     :", resultleft2)
print("resultright2 Ā ∪ B̄    :", resultright2)

# ----- Plot 1: Membership functions A and B -----
plt.figure()
plt.plot(U, A, marker="o", label="A (membership)")
plt.plot(U, B, marker="s", label="B (membership)")
plt.title("Fuzzy Sets A and B: Membership Functions")
plt.xlabel("Universe U")
plt.ylabel("Membership μ(x)")
plt.ylim(-0.05, 1.05)
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()

# ----- Plot 2: Result1 ((A ∪ B)_comp = Ā ∩ B̄ ) -----
plt.figure()
plt.plot(U, resultleft1, marker="^", label="(A ∪ B)_comp")
plt.plot(U, resultright1, marker="v", label="Ā ∩ B̄")
plt.title("Result1 ((A ∪ B)_comp = Ā ∩ B̄ )")
plt.xlabel("Universe U")
plt.ylabel("Membership μ(x)")
plt.ylim(-0.05, 1.05)
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()

# ----- Plot 3: Result2 ((A ∩ B)_comp = Ā ∪ B̄ ) -----
plt.figure()
plt.plot(U, resultleft2, marker="^", label="(A ∩ B)_comp")
plt.plot(U, resultright2, marker="v", label="Ā ∪ B̄")
plt.title("Result2 ((A ∩ B)_comp = Ā ∪ B̄ )")
plt.xlabel("Universe U")
plt.ylabel("Membership μ(x)")
plt.ylim(-0.05, 1.05)
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()
```
