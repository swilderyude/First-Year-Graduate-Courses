```python
# Cartesian Product of Two Fuzzy Sets
# A = { (x1, 0.3), (x2, 0.7), (x3, 1.0) }
# B = { (y1, 0.4), (y2, 0.9) }

import numpy as np
import pandas as pd

# Define fuzzy sets
A_labels = ["x1", "x2", "x3"]
B_labels = ["y1", "y2"]
A_values = np.array([0.3, 0.7, 1.0])
B_values = np.array([0.4, 0.9])

# Compute Cartesian product: μ(A×B)(xi, yj) = min(μA(xi), μB(yj))
cartesian_product = np.zeros((len(A_values), len(B_values)))

for i in range(len(A_values)):
    for j in range(len(B_values)):
        cartesian_product[i, j] = min(A_values[i], B_values[j])

# Display result in table form
df = pd.DataFrame(cartesian_product, index=A_labels, columns=B_labels)
print("Fuzzy Cartesian Product A × B (μ = min(μA, μB)):\n")
print(df)
```
