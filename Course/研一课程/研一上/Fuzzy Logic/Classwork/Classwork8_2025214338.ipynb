```python
import numpy as np
import matplotlib.pyplot as plt

# 定义高斯隶属函数
def gaussmf(x, sigma, c):
    return np.exp(-((x - c)**2) / (2 * sigma**2))

# 定义输入范围
x = np.arange(0, 10.1, 0.1)

# 定义高斯函数参数 sigma=0.5, c=5
mu = gaussmf(x, sigma=0.5, c=5)

# 绘图
plt.plot(x, mu, 'b', linewidth=2)
plt.title('Example 4.8: Gaussian Membership Function')
plt.xlabel('x')
plt.ylabel('Membership degree μ(x)')
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 定义三角形隶属函数
def trimf(x, abc):
    a, b, c = abc
    return np.maximum(np.minimum((x - a)/(b - a), (c - x)/(c - b)), 0)

# 定义输入范围
x = np.arange(0, 10.2, 0.2)

# 定义参数 [3, 4, 5]
mu = trimf(x, [3, 4, 5])

# 绘图
plt.plot(x, mu, 'r', linewidth=2)
plt.title('Example 4.9: Triangular Membership Function')
plt.xlabel('x')
plt.ylabel('Membership degree μ(x)')
plt.grid(True)
plt.show()
```
