```python
# Cartesian Product of Two Fuzzy Sets
# I = { (0.8,0.4), (0.9, 0.7), (1.0, 1.0),(1.1,0.8),(1.2,0.6)}
# V = { (30, 0.2), (45, 0.8),(60,1.0),(75,0.9),(90,0.7) }
# C = { (0.5,0.4), (0.6,1),(0.5,0.7)}

import numpy as np
import pandas as pd

# Define fuzzy sets
I_labels = ["0.8", "0.9", "1.0", "1.1", "1.2"]
V_labels = ["30", "45", "60", "75", "90"]
C_labels = ["0.5", "0.6", "0.5"]
I_values = np.array([0.4, 0.7, 1.0, 0.8, 0.6])
V_values = np.array([0.2, 0.8, 1.0, 0.9, 0.7])
C_values = np.array([0.4, 1.0, 0.7])

# Compute Cartesian product: μ(A×B)(xi, yj) = min(μA(xi), μB(yj))
# 计算 P = V × I
P = np.zeros((len(I_values), len(V_values)))

for i in range(len(V_values)):
    for j in range(len(I_values)):
        P[i, j] = min(V_values[i], I_values[j])

# 计算 T = I × C
T = np.zeros((len(I_values), len(C_values)))


for i in range(len(I_values)):
    for j in range(len(C_values)):
        T[i, j] = min(I_values[i], C_values[j])

# Display result in table form
print("P = V × I:")
df_P = pd.DataFrame(P, index=V_labels, columns=I_labels)
print("Fuzzy Cartesian Product V × I (μ = min(μV, μI)):\n")
print(df_P)
print("\n" + "=" * 50 + "\n")

print("T = I × C:")
df_T = pd.DataFrame(T, index=I_labels, columns=C_labels)
print("Fuzzy Cartesian Product I × C (μ = min(μI, μC)):\n")
print(df_T)
```
