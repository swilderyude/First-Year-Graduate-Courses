```python
import numpy as np

# ----- Data -----
U = np.array(["x1", "x2", "x3", "x4", "x5"])  # 论域元素
P = np.array([0.1, 0.2, 0.7, 0.5, 0.4])  # 模糊集p的隶属度
Q = np.array([0.9, 0.6, 0.3, 0.2, 0.8])  # 模糊集q的隶属度

# ----- Operations -----
# A_union_B = np.maximum(A, B)  # A ∪ B
# A_inter_B = np.minimum(A, B)  # A ∩ B
# A_comp = 1 - A  # Ā
# B_comp = 1 - B  # B̄


# ----- α-cut函数定义 -----
def alpha_cut(U, membership, alpha):
    """
    参数：
        U: 元素集合
        membership: 隶属度数组
        alpha: 截取阈值 (0 <= alpha <= 1)
    返回：
        满足 μ(x) >= α 的元素集合
    """
    return U[membership >= alpha]


# ----- 计算 -----
P_comp = 1 - P
P_union_Q = np.maximum(P, Q)
P_inter_Q = np.minimum(P, Q)
ruselt1 = Q_union_P_comp = np.maximum(Q, P_comp)
result2 = P_union_P_comp = np.maximum(P, P_comp)

# ----- Numeric results -----
print("Universe U       :", U)
print("P                :", P)
print("Q                :", Q)
print("result(a)        :", alpha_cut(U, P_comp, 0.2))
print("result(b)        :", alpha_cut(U, Q, 0.3))
print("result(c)        :", alpha_cut(U, P_union_Q, 0.5))
print("result(d)        :", alpha_cut(U, P_inter_Q, 0.4))
print("result(e)        :", alpha_cut(U, ruselt1, 0.8))
print("result(f)        :", alpha_cut(U, result2, 0.2))
```

```python
import numpy as np

# ----- Data -----
U = np.array(["x1", "x2", "x3", "x4"])  # 论域元素
A = np.array([0.9, 0.5, 0.2, 0.3])  # 模糊集A的隶属度
B = np.array([0.2, 1, 0.8, 0.4])  # 模糊集B的隶属度
C = np.array([0.1, 0.7, 0.5, 0.6])  # 模糊集C的隶属度
# ----- Operations -----
# A_union_B = np.maximum(A, B)  # A ∪ B
# A_inter_B = np.minimum(A, B)  # A ∩ B
# A_comp = 1 - A  # Ā
# B_comp = 1 - B  # B̄


# ----- α-cut函数定义 -----
def alpha_cut(U, membership, alpha):
    """
    参数：
        U: 元素集合
        membership: 隶属度数组
        alpha: 截取阈值 (0 <= alpha <= 1)
    返回：
        满足 μ(x) >= α 的元素集合
    """
    return U[membership >= alpha]


# ----- 计算 -----
A_comp = 1 - A
B_comp = 1 - B
C_comp = 1 - C

# ----- Numeric results -----
print("Universe U       :", U)
print("A                :", A)
print("B                :", B)
print("C                :", C)
print("(A)0.6           :", alpha_cut(U, A, 0.6))
print("(B)1.0           :", alpha_cut(U, B, 1.0))
print("(C)0.3           :", alpha_cut(U, C, 0.3))
print("(A_comp)0.2      :", alpha_cut(U, A_comp, 0.2))
print("(B_comp)0.8      :", alpha_cut(U, B_comp, 0.8))
print("(C_comp)0.5      :", alpha_cut(U, C_comp, 0.5))
```
